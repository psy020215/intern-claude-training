```python
# Pipeline scaffold — generated with plan mode, annotated by you.
# If any cell holds shell commands (bowtie2, samtools, fastq-dump, meme...),
# make its FIRST line %%bash — otherwise Jupyter runs them as Python and errors.
# Pure-Python analysis cells (pandas, Biopython, matplotlib) need no magic.

# --- Step A: 데이터 필터링 ---
# Module4와 동일하게 iron-replete 사이트를 필터링한 뒤,
# S/N ratio 상위 25%(df_high)와 상위 10%(df_top_decile)를 각각 분리한다.
import pandas as pd

TABLE = "../data/reference/ncomms5910_supp_data1_fur_binding_sites.xlsx"
df = pd.read_excel(TABLE, header=1)
df = df.dropna(subset=["Peak"])
df = df[df["Binding Conditiona"].str.contains("R")]

sn_threshold = df["S/N ratio"].quantile(0.75)
df_high = df[df["S/N ratio"] >= sn_threshold]

df["sn_decile"] = pd.qcut(df["S/N ratio"], 10, labels=False, duplicates="drop")
df_top_decile = df[df["sn_decile"] == df["sn_decile"].max()]

# --- Step B: 서열 추출 ---
# df_high, df_top_decile 각각에 대해 Module4와 동일한 방식
# (중심좌표 ±50bp, strand 처리)으로 새 FASTA(fur_sites_high_sn_for_meme.fasta,
# fur_sites_top_decile_for_meme.fasta)를 만든다.
from Bio import SeqIO

GENOME = "../data/reference/NC_000913.3.fasta"
FLANK = 50
genome = SeqIO.read(GENOME, "fasta")
genome_len = len(genome.seq)

def write_fasta(subset_df, out_path):
    records = []
    for _, row in subset_df.iterrows():
        start, end = row["ChIP-exo Start"], row["ChIP-exo End"]
        center = round((start + end) / 2)
        region_start, region_end = center - FLANK, center + FLANK
        if region_start < 1 or region_end > genome_len:
            continue
        seq = genome.seq[region_start - 1:region_end]
        if "strand" in subset_df.columns and row["strand"] == "-":
            seq = seq.reverse_complement()
        header = f"{row['Peak']}_{row['Transcription Unit']}_{center}"
        records.append(f">{header}\n{seq}\n")
    with open(out_path, "w") as f:
        f.writelines(records)

write_fasta(df_high, "../data/reference/fur_sites_high_sn_for_meme.fasta")
write_fasta(df_top_decile, "../data/reference/fur_sites_top_decile_for_meme.fasta")

""" --- Step C: MEME 재실행 ---
새로 만든 두 FASTA 각각에 대해 Module4와 동일한 MEME 플래그로 motif를 다시 탐색한다."""

#%%bash
#/opt/meme/bin/meme ../data/reference/fur_sites_high_sn_for_meme.fasta \
#-dna -revcomp -mod zoops -nmotifs 3 -minw 15 -maxw 25 \
#-oc ../data/reference/meme_fur_high_sn_out
#/opt/meme/bin/meme ../data/reference/fur_sites_top_decile_for_meme.fasta \
#-dna -revcomp -mod zoops -nmotifs 3 -minw 15 -maxw 25 \
#-oc ../data/reference/meme_fur_top_decile_out

# 문헌상 Fur box(GATAATGATAATCATTATC)를 tomtom 비교용 PWM(furbox_query.meme)으로 변환한 뒤,
# 전체 143개 / 상위 25% / 상위 10% 각 MEME 결과와 tomtom으로 비교한다.
#%%bash
#opt/meme/bin/tomtom -no-ssc -thresh 1.0 -oc ../data/reference/tomtom_full_out \
#../data/reference/furbox_query.meme ../data/reference/meme_fur_out/meme.txt
#/opt/meme/bin/tomtom -no-ssc -thresh 1.0 -oc ../data/reference/tomtom_high_sn_out \
#../data/reference/furbox_query.meme ../data/reference/meme_fur_high_sn_out/meme.txt
#/opt/meme/bin/tomtom -no-ssc -thresh 1.0 -oc ../data/reference/tomtom_top_decile_out \
#../data/reference/furbox_query.meme ../data/reference/meme_fur_top_decile_out/meme.txt

# --- Step D: 결과 비교 ---
# 세 그룹(전체 143개 / 상위 25% / 상위 10%)의 top motif(MEME-1)를 tomtom p-value로
# 문헌 Fur box와 비교하여, S/N ratio가 높을수록 Fur box에 더 가까운 motif가 나오는지 확인한다.
```
